```python
from langchain_community.document_loaders import CharactereTextSplitter
```

```python
text = "Cricket is one of the most popular sports in the world, especially in countries like India, Australia, England, Pakistan, and South Africa. It is played between two teams of eleven players each on a large circular or oval-shaped ground. The game is played with a bat and a ball, and the main objective is to score more runs than the opposing team. Cricket is enjoyed by people of all ages because it combines skill, strategy, teamwork, and sportsmanship"
```

```python
loader = CharactereTextSplitter(text)
```

```python
docs = loader.load()
```

```python
splitter = CharactereTextSplitter(
    chunk_size=100,
    chunk_overlap=0.5,
    separator=" ",
)
```

```python
result = splitter.split(docs)
```

```python
print(result)
```
